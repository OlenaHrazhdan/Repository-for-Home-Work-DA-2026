11. Є така штука як сортування значень категоріальної змінної. Нам це зараз стане в нагоді, адже рівні англійської можна посортувати.

```python
df['Знання англійської мови'] = pd.Categorical(df['Знання англійської мови'],
                      categories=['Elementary','Pre-Intermediate','Intermediate','Upper-Intermediate','Advanced'],
                      ordered=True)

df['Ваш тайтл'] = pd.Categorical(df['Ваш тайтл'],
                      categories=['Intern/Trainee', 'Junior', 'Немає тайтлу', 'Middle', 'Senior', 'Tech Lead', 'Team Lead', 'Architect', 'Head', 'Manager'],
                      ordered=True)
```

13. Виведіть результат методу describe() по колонці розміру ЗП для топ 5 за частотою зустрічання в даних областей (В якій області ви зараз живете?). Зробіть висновки, де вигідніше жити з точки зору ЗП і чи суттєва різниця на Ваш погляд для прийняття рішення про переїзд?

```python
region_col = "В якій області ви зараз живете?"
salary_col = "Зарплата у $$$ за місяць, лише ставка після сплати податків"

# 1. Знаходимо топ-5 областей за кількістю респондентів
top5_regions = df[region_col].value_counts().head(5).index

# 2. Фільтруємо датафрейм та розраховуємо describe() для колонки зарплати
df[df[region_col].isin(top5_regions)].groupby(region_col)[salary_col].describe().round(2)
```

14. Як зростає зарплата з роками? Для цього нам треба відформатувати колонку "Загальний стаж роботи за спеціальністю" аналогічно до того, як ми це робили для колонки 'Знання англійської мови' і потім вивести медіанні значення для кількості років стажу.

```python
exp_col = "Загальний стаж роботи за спеціальністю"
salary_col = "Зарплата у $$$ за місяць, лише ставка після сплати податків"

print(df[exp_col].unique())
```

```python
exp_order = [
    'Менше як 3 місяці',
    '3 місяці',
    'Пів року',
    '1 рік',
    '1,5 року',
    '2 роки',
    '3 роки',
    '4 роки',
    '5 років',
    '6 років',
    '7 років',
    '8 років',
    '9 років',
    '10 років',
    '11 років',
    '12 років',
    '13 років',
    '14 років',
    '15 і більше років'
]

# 1. Створюємо категоріальний тип із заданим порядком
df[exp_col] = pd.Categorical(df[exp_col], categories=exp_order, ordered=True)

# 2. Розраховуємо медіанну ЗП для кожного рівня стажу
salary_by_exp = (
    df.groupby(exp_col, observed=False)[salary_col]
    .median()
    .reset_index()
)

# 3. Виводимо результат із градієнтом
salary_by_exp.style.background_gradient(cmap="Blues", subset=[salary_col]).format({salary_col: "${:,.0f}"})
```

```python
df["Ваш вік"].describe().round()
```

```python
# 1. Задаємо межі інтервалів та їхні назви
bins = [0, 20, 25, 35, 100]
labels = ['до 20', '21-25', '26-35', '36+']

# 2. Створюємо нову категоріальну колонку 'Вікова група'
df['Вікова група'] = pd.cut(df['Ваш вік'], bins=bins, labels=labels, right=True)

# 3. Перевіряємо розподіл респондентів за групами
df['Вікова група'].value_counts().sort_index()
```

```python
df.groupby(["Ваша стать", "Вікова група"])["Ваш тайтл"].value_counts()
```

```python
# 1. Формуємо зведену таблицю
pivot_df = df.pivot_table(
    index=["Ваша стать", "Вікова група"],
    columns="Ваш тайтл",
    aggfunc="size",
    fill_value=0
)

# 2. Застосовуємо градієнтне забарвлення
pivot_df.style.background_gradient(cmap="Blues", axis=None).format("{:,}")
```
