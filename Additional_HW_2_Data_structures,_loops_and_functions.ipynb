4.Напишіть функцію, яка для заданого цілого числа виводить на екран наступну фігуру. Підказка - тут треба використати вкладені цикли for. Наприклад, для `n=5` фігура матиме наступний вигляд:

```
*
* *
* * *
* * * *
* * * * *
* * * *
* * *
* *
*
```

Запустіть фукнцію для значень n: 3, 5, 10.

````python
def draw_zirka (n):
  for i in range (1, n+1):
    print ('* '*i)
  for i in range (n-1, 0, -1):
    print ('* '*i)

draw_zirka (3)
````

````python
draw_zirka (10)
````

````python
def draw_pattern(n):                      # Варіант з вкладеними циклами for
   for i in range(1, n + 1):              # 1. Верхня частина фігури (включно з n)
      for j in range(i):
          print('*', end=' ')
      print()                             # Перехід на новий рядок після завершення рядка зірочок

   for i in range(n - 1, 0, -1):          # 2. Нижня частина фігури (від n-1 до 1)
       for j in range(i):
          print('*', end=' ')
       print()                            # Перехід на новий рядок

draw_pattern(5)                           # Викликаємо функцію для n=5 (або будь-якого іншого числа)
````

Приклад: для наступного входу

numbers = (1, 2, 11, 12, 13, 22, 27, 29)

Функція має повернути результат: `3, 5`.

````python
numbers_1 = {1, 2, 11, 12, 13, 22, 27, 29}
numbers_2 = {1, 22, 412, 312, 73, 3421}
numbers_3 = {12, 11, 94, 722, 0, 452}
````

````python
def par_nonpar (numbers):
  a=0
  b=0

  for i in numbers:
    if i%2==0:
      a+=1
    else:
      b+=1
  # return (a, b)
  print (f'Кількість парних чисел : {a}')
  print (f'Кількість непарних : {b}')
````

````python
# 1. Оголошуємо набори даних
numbers_1 = {1, 2, 11, 12, 13, 22, 27, 29}
numbers_2 = {1, 22, 412, 312, 73, 3421}
numbers_3 = {12, 11, 94, 722, 0, 452}

# 2. Функція підрахунку (вертає два числа)
def par_nonpar(numbers):
    a = 0  # лічильник парних
    b = 0  # лічильник непарних

    for i in numbers:
        if i % 2 == 0:
            a += 1
        else:
            b += 1

    return a, b

# 3. Об'єднуємо всі набори в список і проходимо по них у циклі
all_numbers = [numbers_1, numbers_2, numbers_3]

for num_set in all_numbers:
    # Отримуємо результати з функції
    even_count, odd_count = par_nonpar(num_set)

    # Виводимо форматоване повідомлення
    print(f"Кількість парних чисел : {even_count}")
    print(f"Кількість непарних : {odd_count}")
    print("-" * 25)  # Розділювач для красивого виводу
````

7.Уявіть, що Ви створюєте пригодницьку гру і використовуєте для зберігання предметів гравця словник, у якому ключі - це назви предметів, значення - кількість одиниць кожної із речей. Наприклад, словник може виглядати так: things = {'ключ': 3, 'булава': 1, 'монета': 24, 'ніж': 1, 'горішки': 10, 'бронежелет': 0}. Створіть фукнцію, яка виводить повідомлення про усі речі гравця у такому вигляді:

```
У гравця є:
3 ключ
1 булава
24 монета
1 ніж
10 горішки
0 бронежелет
Загальна кількість речей: 39
```

Викличіть фукнцію для наступних гравців:

````python
player_1 = {'ключ': 3, 'булава': 1, 'монета': 24, 'ніж': 1, 'горішки': 10, 'бронежелет': 0}
player_2 = {'ключ': 0, 'булава': 2, 'монета': 4, 'ніж': 3, 'горішки': 4, 'бронежелет': 1}
player_3 = {'ключ': 0, 'булава': 0, 'монета': 30, 'ніж': 0, 'горішки': 2, 'бронежелет': 2}
````

````python
def things(player):
    print("У гравця є:")                   # Заголовок "У гравця є:" повинен виводитися один раз до початку циклу, а не на кожній ітерації.
                                           # Виводимо кожен предмет та його кількість
    for item, count in player.items():     # Метод .items() бере кожен елемент словника і віддає його у вигляді пари (кортежу): (ключ, значення).
        print(f"{count} {item}")

    total_count = sum(player.values())     # Рахуємо загальну суму та виводимо її після циклу
    print(f"Загальна кількість речей: {total_count}")
````

````python
things (player_1)
````

````python
things (player_2)
````

````python
things (player_3)
````

````python
 # Запуск функції для всіх списків у циклі
all_players = [player_1, player_2, player_3]

for i in all_players:
      things (i)
      print ("-"*30)
````
