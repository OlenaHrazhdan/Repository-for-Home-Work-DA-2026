1. Дослідимо, який філіал супермаркету ('Branch') є найприбутковішим. Для цього знайдіть сумарний прибуток за кожним філіалом і виявіть, який філіал має найвищий.

2. В якому місті знайходиться філіал з найвищим прибутком? Може в тому місці нам розмітисти ще один магазин.  
Знайдіть відповідь за допомогою функціоналу Pandas.

3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.
Ця таблиця допоможе проаналізувати, в якій філії купують найбільше товарів кожної з категорій.

```python
invoices_by_category = pd.pivot_table(df, index="Branch", columns="Product line", values="Invoice ID", aggfunc="size")
```

1. Яка категорія товарів найбільш затребувана у філії міста Naypyitaw (за кількістю інвойсів)?

```python
df[df['City'] == 'Naypyitaw'].groupby('Product line')['Invoice ID'].count().sort_values(ascending=False).head(1)
```

Найбільше користується попитом у місті Naypyitaw категорія товарів Food and beverages. Її можні масштабувати і розширювати.

2. Яка категорія товарів приносить найменший прибуток у філії міста Yangon?

```python
df[df['City'] == 'Yangon'].groupby('Product line')['gross income'].sum().sort_values().head(1)
```

Найменший дохід у місті Yangon приносить категорія товарів Healt and beauty. Потрібно переглянути асортимент наявний у цьому магазині, також ввести якісь промо-акції для стимулювання продажів даної категорії.

3. Який тип користувачів приносить найбільший дохід (постійні члени чи звичайні), який їх розподіл?

```python
df.groupby("Customer type").agg({'gross income': 'sum', 'Invoice ID' : 'count'})
```

Постійні члени приносять трохи більший дохід і здійснюють трохи більше половини покупок. Можна зробити висновок, що програма лояльності працює. Потрібно залучати і простих користувачів в неї.
